##### Define inputs, parameters:

```python
# Input file:
input_path = os.path.join('/', 'mnt', 'smb', 'locker', 'issa-locker', 'users', 'Dan', 'code', 'ws', 'results', 'run759', 'ae_iterate_beta_reconstruction.pickle')

# Output parameters:
save_output = True
```

##### Load pre-saved results, extract relevant objects:

```python
results = pickle.load(open(input_path, 'rb'))
geo_df = results['geo_df']
ae_df = geo_df[geo_df.model_type=='autoencoder']
pred_df = geo_df[geo_df.model_type=='prediction']
```

##### Compute mean parallelism for each model type:

```python
T = np.mean(np.mean(ae_df[ae_df.layer=='hidden']['parallelism']))
P = np.mean(np.mean(pred_df[pred_df.layer=='hidden']['parallelism']))
```

##### Plot:

```python
f = plt.figure()
plt.bar([0,1], [T,P])
plt.xticks([0,1], labels=['entangler model hidden layer', 'prediction model hidden layer'])
plt.ylabel('Parallelism score (PS)')
plt.title('Hidden layer parallelism score')
```
